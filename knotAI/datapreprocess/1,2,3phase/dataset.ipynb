```python

import kagglehub
import shutil
import os

# 1. 구글 드라이브 내 최종 저장 경로 지정 (예: 10knots 폴더)
drive_path = "/content/drive/MyDrive/knot_dataset/10knots/Knot_origianal_data"
os.makedirs(drive_path, exist_ok=True)

# 2. kagglehub를 이용한 데이터 다운로드 (기본 캐시 폴더에 저장됨)
print("데이터 다운로드를 시작합니다...")
cache_path = kagglehub.dataset_download("josephcameron/10knots")
print("임시 캐시 저장 경로:", cache_path)

# 3. 다운로드된 파일들을 구글 드라이브 경로로 복사 (Copy)
print("구글 드라이브로 파일을 복사하는 중...")
for item in os.listdir(cache_path):
    source = os.path.join(cache_path, item)
    destination = os.path.join(drive_path, item)
    
    # 폴더(Directory)인 경우 통째로 복사
    if os.path.isdir(source):
        shutil.copytree(source, destination, dirs_exist_ok=True)
    # 파일(File)인 경우 단일 복사
    else:
        shutil.copy2(source, destination)

print("구글 드라이브 최종 저장 완료:", drive_path)
```

```python
import os
import shutil
import time

def safe_copy_tree(source, destination, max_retries=3):
    failed = []
    copied = 0

    for dirpath, dirnames, filenames in os.walk(source):
        rel = os.path.relpath(dirpath, source)
        dst_dir = os.path.join(destination, rel) if rel != "." else destination

        for attempt in range(max_retries):
            try:
                os.makedirs(dst_dir, exist_ok=True)
                break
            except OSError as e:
                if attempt == max_retries - 1:
                    failed.append((dirpath, str(e)))
                time.sleep(2)

        for fname in filenames:
            src_file = os.path.join(dirpath, fname)
            dst_file = os.path.join(dst_dir, fname)

            for attempt in range(max_retries):
                try:
                    shutil.copy2(src_file, dst_file)
                    copied += 1
                    break
                except OSError as e:
                    if attempt == max_retries - 1:
                        failed.append((src_file, str(e)))
                    time.sleep(2)

    print(f"복사 완료: {copied}개")
    print(f"실패: {len(failed)}개")
    if failed:
        print("실패 목록 (앞 5개):")
        for f, err in failed[:5]:
            print(f"  {f} → {err}")
    return failed

# 사용
failed = safe_copy_tree(cache_path, drive_path)
```

```python
import os
import shutil

def flatten_below_depth(root_dir, keep_depth=0, dry_run=True):
    root_dir = os.path.abspath(root_dir)
    
    # 1. 경로 존재 여부 검사 (로컬 경로 오류 잡기)
    if not os.path.exists(root_dir):
        print(f"❌ 경로를 찾을 수 없습니다: {root_dir}")
        return

    moves = []

    for dirpath, dirnames, filenames in os.walk(root_dir, topdown=False):
        rel_path = os.path.relpath(dirpath, root_dir)
        depth = 0 if rel_path == "." else len(rel_path.split(os.sep))

        if depth > keep_depth and filenames:
            parts = rel_path.split(os.sep)
            target_rel = os.sep.join(parts[:keep_depth])
            target_dir = os.path.join(root_dir, target_rel) if target_rel else root_dir

            for fname in filenames:
                src = os.path.join(dirpath, fname)
                dst = os.path.join(target_dir, fname)

                if os.path.exists(dst) or any(m[1] == dst for m in moves):
                    prefix = "_".join(parts[keep_depth:])
                    base, ext = os.path.splitext(fname)
                    dst = os.path.join(target_dir, f"{prefix}_{base}{ext}")

                moves.append((src, dst))

    print(f"\n[{os.path.basename(root_dir)}] 총 {len(moves)}개 파일 이동 예정 (dry_run={dry_run})")
    
    if len(moves) == 0:
        print(" -> 하위 폴더가 없거나 이미 평탄화가 완료된 상태입니다.")

    for src, dst in moves[:3]:
        print(f"  {src}  →  {dst}")

    if dry_run or not moves:
        return

    for src, dst in moves:
        os.makedirs(os.path.dirname(dst), exist_ok=True)
        shutil.move(src, dst)

    for dirpath, dirnames, filenames in os.walk(root_dir, topdown=False):
        rel_path = os.path.relpath(dirpath, root_dir)
        depth = 0 if rel_path == "." else len(rel_path.split(os.sep))
        if depth > keep_depth:
            try:
                os.rmdir(dirpath)
            except OSError:
                pass

    print("완료.")


a = [
    'Alpine Butterfly Knot', 'Bowline Knot', 'Clove Hitch', 
    'Figure-8 Knot', 'Figure-8 Loop', "Fisherman's Knot", 
    'Flemish Bend', 'Reef Knot', 'Slip Knot', 'Overhand Knot'
]

# ★ 로컬 컴퓨터 환경에 맞는 실제 폴더 경로로 변경하세요!
# 예: Windows -> "C:/Users/내이름/Desktop/knot_dataset/10knots/knot_test_data"
# 예: 상대경로 -> "./knot_dataset/10knots/knot_test_data"
base_path = r"G:\내 드라이브\knot_dataset\10knots\knot_test_data"

for knot_name in a:
    path = os.path.join(base_path, knot_name)
    flatten_below_depth(path, keep_depth=0, dry_run=False)
```

```python
import kagglehub
import shutil
import os

# 1. 기존 구글 드라이브 데이터 완전 삭제
drive_path = "/content/drive/MyDrive/knot_dataset/10knots/Knot_origianal_data"

if os.path.exists(drive_path):
    print(f"기존 폴더 삭제 중: {drive_path}")
    shutil.rmtree(drive_path)
    print("삭제 완료.")

os.makedirs(drive_path, exist_ok=True)

# 2. 캐글에서 데이터 새로 다운로드 (기존 캐시 무시하고 강제 재다운로드)
print("\n캐글에서 데이터 다운로드 중...")
cache_path = kagglehub.dataset_download("josephcameron/10knots", force_download=True)
print("다운로드 완료. 임시 저장 경로:", cache_path)

# 3. 구글 드라이브로 복사 (실패해도 안 멈추고 진행률 표시)
def safe_copy_tree(source, destination):
    failed = []
    copied = 0
    total = sum(len(files) for _, _, files in os.walk(source))
    print(f"\n전체 {total}개 파일 복사 시작...")

    for dirpath, dirnames, filenames in os.walk(source):
        rel = os.path.relpath(dirpath, source)
        dst_dir = os.path.join(destination, rel) if rel != "." else destination
        os.makedirs(dst_dir, exist_ok=True)

        for fname in filenames:
            src_file = os.path.join(dirpath, fname)
            dst_file = os.path.join(dst_dir, fname)
            try:
                shutil.copy2(src_file, dst_file)
                copied += 1
                if copied % 50 == 0:
                    print(f"진행: {copied}/{total}")
            except OSError as e:
                failed.append((src_file, str(e)))

    print(f"\n복사 완료: {copied}개 성공, {len(failed)}개 실패")
    if failed:
        print("실패 목록 (앞 5개):")
        for f, err in failed[:5]:
            print(f"  {f} → {err}")
    return failed

failed = safe_copy_tree(cache_path, drive_path)
print("\n✅ 구글 드라이브 최종 저장 완료:", drive_path)
```
